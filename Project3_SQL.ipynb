## Objective
Use SQL queries to extract business insights from the Customer Sales dataset. This project demonstrates SELECT, WHERE, ORDER BY, GROUP BY, and aggregation functions (COUNT, SUM, AVG).

---

## Step 1: Setup — Load Dataset into SQLite Database

```python
import pandas as pd
import sqlite3
import warnings
warnings.filterwarnings('ignore')

# Load cleaned CSV into pandas
df = pd.read_csv('cleaned_customer_data.csv')

# Create SQLite database in memory
conn = sqlite3.connect('customer_sales.db')

# Load dataframe into SQL table
df.to_sql('customer_sales', conn, if_exists='replace', index=False)

print('=== DATABASE SETUP COMPLETE ===')
print(f'Table: customer_sales')
print(f'Total Records: {len(df)}')
print(f'Columns: {list(df.columns)}')

# Helper function to run SQL and show results
def run_sql(query):
    result = pd.read_sql_query(query, conn)
    return result
```

## Step 2: Basic SELECT — View the Data

```python
print('=== QUERY 1: View First 10 Records ===')
query1 = '''
SELECT *
FROM customer_sales
LIMIT 10
'''
result1 = run_sql(query1)
print(result1.to_string(index=False))
```

## Step 3: SELECT Specific Columns

```python
print('=== QUERY 2: Customer Name, City and Revenue Only ===')
query2 = '''
SELECT Customer_Name, City, Product_Category, Revenue
FROM customer_sales
ORDER BY Revenue DESC
LIMIT 10
'''
result2 = run_sql(query2)
print(result2.to_string(index=False))
print()
print('Insight: Top 10 highest revenue customers shown above.')
```

## Step 4: WHERE Clause — Filter Data

```python
print('=== QUERY 3: Active Customers Only ===')
query3 = '''
SELECT Customer_ID, Customer_Name, City, Revenue, Status
FROM customer_sales
WHERE Status = 'Active'
ORDER BY Revenue DESC
'''
result3 = run_sql(query3)
print(result3.to_string(index=False))
print(f'\nTotal Active Customers: {len(result3)}')
```

```python
print('=== QUERY 4: High Value Customers (Revenue > 2000) ===')
query4 = '''
SELECT Customer_ID, Customer_Name, City, Product_Category, Revenue
FROM customer_sales
WHERE Revenue > 2000
ORDER BY Revenue DESC
'''
result4 = run_sql(query4)
print(result4.to_string(index=False))
print(f'\nHigh Value Customers: {len(result4)}')
```

```python
print('=== QUERY 5: Customers from Lahore who bought Electronics ===')
query5 = '''
SELECT Customer_ID, Customer_Name, City, Product_Category, Revenue
FROM customer_sales
WHERE City = 'Lahore'
AND Product_Category = 'Electronics'
ORDER BY Revenue DESC
'''
result5 = run_sql(query5)
print(result5.to_string(index=False))
print(f'\nTotal: {len(result5)} customers')
```

## Step 5: ORDER BY — Sort Results

```python
print('=== QUERY 6: All Customers Sorted by Age (Youngest First) ===')
query6 = '''
SELECT Customer_Name, Age, City, Revenue
FROM customer_sales
ORDER BY Age ASC
LIMIT 10
'''
result6 = run_sql(query6)
print(result6.to_string(index=False))

print()
print('=== QUERY 7: Top 5 Highest Revenue Customers ===')
query7 = '''
SELECT Customer_Name, City, Product_Category, Revenue
FROM customer_sales
ORDER BY Revenue DESC
LIMIT 5
'''
result7 = run_sql(query7)
print(result7.to_string(index=False))
```

## Step 6: GROUP BY — Aggregate by Category

```python
print('=== QUERY 8: Total Revenue by Product Category (GROUP BY) ===')
query8 = '''
SELECT
    Product_Category,
    COUNT(*) AS Total_Customers,
    SUM(Revenue) AS Total_Revenue,
    AVG(Revenue) AS Avg_Revenue,
    MIN(Revenue) AS Min_Revenue,
    MAX(Revenue) AS Max_Revenue
FROM customer_sales
GROUP BY Product_Category
ORDER BY Total_Revenue DESC
'''
result8 = run_sql(query8)
result8 = result8.round(2)
print(result8.to_string(index=False))
print()
print('Insight: Clothing leads in total revenue. Food has highest average revenue per customer.')
```

```python
print('=== QUERY 9: Revenue Analysis by City (GROUP BY) ===')
query9 = '''
SELECT
    City,
    COUNT(*) AS Total_Customers,
    SUM(Revenue) AS Total_Revenue,
    ROUND(AVG(Revenue), 2) AS Avg_Revenue
FROM customer_sales
GROUP BY City
ORDER BY Total_Revenue DESC
'''
result9 = run_sql(query9)
print(result9.to_string(index=False))
print()
print('Insight: Lahore has the most customers and highest total revenue.')
```

```python
print('=== QUERY 10: Active vs Inactive — Revenue Comparison (GROUP BY) ===')
query10 = '''
SELECT
    Status,
    COUNT(*) AS Customer_Count,
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(AVG(Revenue), 2) AS Avg_Revenue
FROM customer_sales
GROUP BY Status
ORDER BY Total_Revenue DESC
'''
result10 = run_sql(query10)
print(result10.to_string(index=False))
```

## Step 7: COUNT, SUM, AVG — Aggregation Functions

```python
print('=== QUERY 11: Overall Business Summary ===')
query11 = '''
SELECT
    COUNT(*) AS Total_Customers,
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(AVG(Revenue), 2) AS Avg_Revenue_Per_Customer,
    ROUND(AVG(Age), 1) AS Avg_Customer_Age,
    MIN(Revenue) AS Lowest_Purchase,
    MAX(Revenue) AS Highest_Purchase
FROM customer_sales
'''
result11 = run_sql(query11)
print(result11.to_string(index=False))
```

```python
print('=== QUERY 12: Customer Count by City and Category ===')
query12 = '''
SELECT
    City,
    Product_Category,
    COUNT(*) AS Customer_Count,
    ROUND(SUM(Revenue), 2) AS Total_Revenue
FROM customer_sales
GROUP BY City, Product_Category
ORDER BY City, Total_Revenue DESC
'''
result12 = run_sql(query12)
print(result12.to_string(index=False))
```

## Step 8: Advanced Filtering with WHERE + GROUP BY

```python
print('=== QUERY 13: Categories with Average Revenue Above 2000 ===')
query13 = '''
SELECT
    Product_Category,
    COUNT(*) AS Customers,
    ROUND(AVG(Revenue), 2) AS Avg_Revenue
FROM customer_sales
GROUP BY Product_Category
HAVING AVG(Revenue) > 2000
ORDER BY Avg_Revenue DESC
'''
result13 = run_sql(query13)
print(result13.to_string(index=False))
print()
print('Insight: Categories where avg revenue exceeds PKR 2000.')
```

```python
print('=== QUERY 14: Young High-Value Customers (Age < 30 AND Revenue > 2000) ===')
query14 = '''
SELECT Customer_Name, Age, City, Product_Category, Revenue
FROM customer_sales
WHERE Age < 30
AND Revenue > 2000
ORDER BY Revenue DESC
'''
result14 = run_sql(query14)
print(result14.to_string(index=False))
print(f'\nTotal: {len(result14)} young high-value customers')
```

## Step 9: Final Business Summary Query

```python
print('=== QUERY 15: Complete Business Intelligence Report ===')
query15 = '''
SELECT
    City,
    Product_Category,
    Status,
    COUNT(*) AS Customers,
    ROUND(AVG(Revenue), 2) AS Avg_Revenue,
    ROUND(SUM(Revenue), 2) AS Total_Revenue
FROM customer_sales
GROUP BY City, Product_Category, Status
ORDER BY Total_Revenue DESC
LIMIT 10
'''
result15 = run_sql(query15)
print(result15.to_string(index=False))

# Close connection
conn.close()
print()
print('Database connection closed.')
```
